# SG Job Data: Data Cleaning


```python
import pandas as pd
```

```python
#load SGJobData.csv into the data folder
df = pd.read_csv('SGJobData.csv')
```

## 1. Load and inspect the data

```python
df.shape                    # 1. How big is your sample? (rows, columns)
df.info()                   # 2. What columns do you have, and what types are they?
df.describe()     # 3. Pick one numeric column — what's typical, what's extreme?
```

## 2. Cleaning up Data Types and Columns

```python
date_columns = [
    "metadata_newPostingDate",
    "metadata_expiryDate",
    "metadata_originalPostingDate"
]

for col in date_columns:
    df[col] = pd.to_datetime(df[col], dayfirst=True)

df.dtypes
```

```python
df.isna().sum()
```

```python
key_columns = [
    "categories",
    "employmentTypes",
    "metadata_jobPostId",
    "positionLevels",
    "postedCompany_name",
    "salary_type",
    "title",
]
df[key_columns].isna().all(axis=1).sum()
```

```python
df = df.dropna(subset=key_columns, how='all')
```

```python
df = df.drop(columns=["occupationId"])
df.shape
```

## 3. Inspect other string / categorical columns

```python
columns_to_check = [
    "employmentTypes",
    "positionLevels",
    "salary_type",
    "status_id"
]

for col in columns_to_check:
    print(f"\nValue counts for {col}:")
    print(df[col].value_counts())
```

```python
df = df.drop(columns=["status_id","salary_type"])
df.shape
```

```python
#identifying duplicates

number_of_duplicates =df.duplicated(['metadata_jobPostId']).sum()
display(number_of_duplicates)
```

## 4. Explore and clean salary data

```python
salary_df = df[
    ["title", "salary_minimum", "salary_maximum", "average_salary","positionLevels","employmentTypes"]
].copy()

salary_df = salary_df.sort_values(
    by="salary_maximum",
    ascending=True
)

salary_df.head(5)
```

```python
#looking at the columns in salary minimum and maximum to see if there are any outliers
salary_df[[
    "salary_minimum",
    "salary_maximum",
    "average_salary"
]].describe(percentiles=[0.01,0.05, 0.10, 0.25, 0.5, 0.75, 0.90, 0.95, 0.99])
```

```python
salary_df[(salary_df['salary_maximum'] < 1000) | (salary_df['salary_maximum'] > 20000)].shape
# Filter maximum salaries below $1,000 (1st percentile) and above $20,000 (99th percentile).
# Rationale:
# - Use maximum salary as the filtering variable, as some listings have a minimum salary of $1 ('invalid') but a 'valid' maximum salary (e.g. $5000)
# - Filtering based on minimum salary could therefore exclude otherwise useful listings.
```

```python
salary_df[(salary_df['salary_maximum'] < 1000) | (salary_df['salary_maximum'] > 20000)].sort_values(by='salary_maximum').head(10)
```

```python
salary_df[(salary_df['salary_maximum'] < 1000) | (salary_df['salary_maximum'] > 20000)].sort_values(by='salary_maximum').tail(10)
```

```python
salary_df.shape

#confirming how many rows are where maximum salary < $1000 and >$20,000 to see the implications of removing these rows
```

```python
filtered_salary = salary_df[(salary_df['salary_maximum'] >= 1000) & (salary_df['salary_maximum'] <= 20000)]
filtered_salary.describe()
filtered_salary.sort_values(by='average_salary').head(10)

# Removed observations below the 1st and above the 99th percentiles

# Rationale:
# - Due to time constraints, outliers were not manually validated against original listings to determine if values were erroneous (e.g. keyed in extra zeros) or legitimate requirements
# - These observations represent a small proportion of the dataset (18k out of 1 million rows)
# - Removing them reduces the potential distortion of salary summaries while retaining main 98% of observations.
```

```python

# Applying this step to the main dataframe

df = df[(df['salary_maximum'] >= 1000) & (df['salary_maximum'] <= 20000)]
df.shape
```

## 5. Group employment types and position levels

```python
#mapping employmentTypes
emp_mapping = { "Permanent": "Full Time & Perm", 
               "Full Time": "Full Time & Perm", 
               "Contract": "Contract & Temp", 
               "Temporary": "Contract & Temp",
               "Part Time": "Part-Time, Flexi, Freelance",
               "Freelance": "Part-Time, Flexi, Freelance", 
               "Flexi-work": "Part-Time, Flexi, Freelance", 
               "Internship/Attachment": "Intern" } 
df["employmentTypes_grouped"] = df["employmentTypes"].map(emp_mapping) 
print(df["employmentTypes_grouped"].value_counts())
```

```python
# mapping positionLevels

pos_mapping = {
    "Fresh/entry level": "Entry & Junior", 
    "Junior Executive": "Entry & Junior",
    "Non-executive": "Entry & Junior", 
    
    "Executive": "Mid-Level",
    "Senior Executive": "Mid-Level",
    "Professional": "Mid-Level",
    
    "Manager": "Mgmt",
    "Middle Management": "Mgmt",
    "Senior Management": "Mgmt"}

df["positionLevels_grouped"] = df["positionLevels"].map(pos_mapping)
print(df["positionLevels_grouped"].value_counts())
```

## 6. Investigate and filter minimum years of experience

```python
#created new dataframe to investigate just minimumYearsExperience
years_df = df[["title", "minimumYearsExperience","positionLevels_grouped","employmentTypes_grouped"]].copy()
display(years_df.sort_values(by="minimumYearsExperience",ascending=False).head(10))
```

```python
years_df.describe(percentiles=[0.01,0.05, 0.10, 0.25, 0.5, 0.75, 0.90, 0.95, 0.99])
```

```python
years_df[years_df["minimumYearsExperience"] > 10].shape[0]
# Using this command to investigate the years of experience required by postings requring >10 years of minimum experience
```

```python
# Peaks in positions requring 20, 25 and 30 years of experience. 
# Investigating positions requring 30 years of experience first
years_df[(years_df['minimumYearsExperience'] <= 30)].sort_values(by='minimumYearsExperience',ascending=False).head(10)
```

##### based on for positions requring 30 years experience, it appears there are anomalies e.g. Entry & Junior levels requring above 30 years

### Compare the original and filtered experience distributions

```python
filtered_years = years_df[(years_df['minimumYearsExperience'] <= 25)]

print("original")
years_df['minimumYearsExperience'].describe()
```

```python
print("filtered")
display(filtered_years['minimumYearsExperience'].describe())
```

## 7. Apply the experience filter to the main dataframe

```python
df = df[(df['minimumYearsExperience'] <= 25)] 
df.shape

# Decision: Exclude postings requiring more than 25 years of experience.

# Rationale:
# - Only 122 records (a small proportion of the dataset) exceeded this threshold.
# - Similar to "salary" records could not be individually reviewed against their original job postings to verify their accuracy.
# - Excluding them was a pragmatic approach to limit the influence of potentially anomalous values 
```
